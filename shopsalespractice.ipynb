```python
#3 find the top 5 products by reveue and visualize them 
df.head()
top=df.groupby("product")["total_sales"].sum().sort_values(ascending=False)
top5=top.head()
plt.barh(top5.index,top5.values,label="top 5 poduct by revenue")
plt.xlabel("revenue")
plt.ylabel("product")
plt.title("horizontal bar chart")
plt.legend()
plt.show()
```

```python
#4 analyze category performance using sales, quantity and a chart

top=df.groupby("category")["total_sales"].sum()
plt.barh(top.index,top.values,label="category wise revenue")
plt.xlabel("revenue")
plt.ylabel("category")
plt.title("horizontal bar chart")
plt.legend()
plt.show()
```

```python
top = df.groupby("category")[["total_sales", "quantity"]].sum()

print(top)

# Sales chart
plt.bar(top.index, top["total_sales"], label="sales")

for i, value in enumerate(top["total_sales"]):
    plt.text(i, value, value, ha="center", va="bottom")

plt.xlabel("category")
plt.ylabel("values")
plt.title("Category Performance in total_sales")
plt.legend()
plt.show()


# Quantity chart
plt.bar(top.index, top["quantity"], label="quantity")
for i , value in enumerate(top["quantity"]):
    plt.text(i, value,value , ha="center", va="bottom")

plt.xlabel("category")
plt.ylabel("value")
plt.title("Category Performance in quantity")
plt.legend()
plt.show()
```

```python
#5 analyze category performance and vusualize the results
df.head()
top = df.groupby("customer_city")["total_sales"].sum()
plt.bar(top.index, top.values)
for i ,value in enumerate(top.values):
    plt.text(i,value,value,ha="center",va="bottom")

plt.xlabel("city")
plt.ylabel("total sales")
plt.title("City Performance")

plt.show()
```

```python
#6 find monthly sales using pandas , calculate statistics with numpy and careate a line chart 
df.head()
totalsale=df.groupby("month")["total_sales"].sum()
print("average :", round(np.mean(totalsale.values),2))
print("minimum :", np.min(totalsale.values))
print("maximum :", np.max(totalsale.values))
month_order = [
    "January", "February", "March",
    "April", "May", "June",
    "July", "August", "September",
    "October", "November", "December"
]
total=totalsale.reindex(month_order).dropna()
plt.plot(total.index,total.values,marker="o")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales")
plt.show()
```

```python
#7 identify high value orders and visualize their distribution.
df.head()
avg = df["total_sales"].mean()

high = df[df["total_sales"] > avg]

#print(high)

plt.hist(high["total_sales"])

plt.xlabel("Total Sales")
plt.ylabel("Number of Orders")
plt.title("Distribution of High-Value Orders")

plt.show()
```

```python
# Q8: Relationship between Quantity and Total Sales

print("Correlation:")
print(df[["quantity", "total_sales"]].corr())

plt.scatter(df["quantity"], df["total_sales"])

plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.title("Relationship Between Quantity and Total Sales")

plt.show()
```

```python
# Q9: Compare Payment Methods by Number of Orders and Total Revenue

payment = df.groupby("payment_method").agg(
    orders=("order_id", "count"),
    revenue=("total_sales", "sum")
)

print(payment)

plt.bar(payment.index, payment["orders"])

plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.title("Number of Orders by Payment Method")

plt.show()

plt.bar(payment.index, payment["revenue"])

plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.title("Total Revenue by Payment Method")

plt.show()
```
